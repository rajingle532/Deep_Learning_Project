# 🌿 Phase 2: Exploratory Data Analysis & Quantifying the Domain Gap
### Domain-Generalized Deep Learning for Real-World Plant Disease Recognition

**Objective:** Empirically and mathematically prove why standard Deep Learning models fail when deployed from controlled laboratories (**PlantVillage**) to in-the-wild agricultural fields (**PlantDoc**).

```python
import json
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image

PROJECT_ROOT = Path("..")
RESULTS_EDA = PROJECT_ROOT / "results" / "eda"
PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"

# Load manifests
df_train = pd.read_csv(PROCESSED_DIR / "train_manifest.csv")
df_val = pd.read_csv(PROCESSED_DIR / "val_manifest.csv")
df_test = pd.read_csv(PROCESSED_DIR / "test_manifest.csv")

print(f"Source Domain Train: {len(df_train):,} images (PlantVillage)")
print(f"Source Domain Val:   {len(df_val):,} images (PlantVillage)")
print(f"Target Domain Test:  {len(df_test):,} images (PlantDoc strictly held-out)")
```

## 1. Class Distribution & Imbalance Analysis across Domains
We compare the frequency of all 28 canonical classes across both domains and compute inverse-frequency class weights:  
$$w_c = \frac{N_{\text{total}}}{C \cdot N_c}$$

```python
# Display generated Class Distribution Comparison Chart
dist_img = Image.open(RESULTS_EDA / "class_distribution_comparison.png")
plt.figure(figsize=(14, 10))
plt.imshow(dist_img)
plt.axis("off")
plt.title("Class Distribution: PlantVillage vs PlantDoc", fontsize=14, fontweight="bold")
plt.show()
```

## 2. Empirical Proof of Covariate Shift (Color Space Profile)
Comparing RGB and HSV channel density distributions reveals why laboratory models collapse in real fields:

```python
# Display Color Space Covariate Shift Curves
color_img = Image.open(RESULTS_EDA / "color_space_domain_shift.png")
plt.figure(figsize=(15, 9))
plt.imshow(color_img)
plt.axis("off")
plt.title("Color Space Covariate Shift Profile", fontsize=14, fontweight="bold")
plt.show()
```

## 3. Image Texture & Sharpness Breakdown (Laplacian Variance)
Evaluating image brightness, contrast, and Laplacian variance (blurriness metric: $\text{Var}(\nabla^2 I)$):

```python
# Display Image Quality Shift Chart
qual_img = Image.open(RESULTS_EDA / "image_quality_shift.png")
plt.figure(figsize=(15, 6))
plt.imshow(qual_img)
plt.axis("off")
plt.title("Image Sharpness & Quality Shift", fontsize=14, fontweight="bold")
plt.show()
```

```python
# Print statistical summary
with open(RESULTS_EDA / "eda_summary.json", "r") as f:
    summary = json.load(f)

stats = summary["image_quality_statistics"]
print("=" * 60)
print("STATISTICAL SUMMARY: DOMAIN GAP PROOF")
print("=" * 60)
print(f"PlantVillage Mean Sharpness (Laplacian Var): {stats['source']['laplacian_var_mean']}")
print(f"PlantDoc Mean Sharpness (Laplacian Var):     {stats['target']['laplacian_var_mean']}")
print(f"Sharpness Drop Ratio: {stats['source']['laplacian_var_mean'] / stats['target']['laplacian_var_mean']:.2f}x")
print("=" * 60)
```
